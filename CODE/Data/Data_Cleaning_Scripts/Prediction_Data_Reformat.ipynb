```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
bridge_data_path = "D:\\Bridge_Data\\CSV\\Cleaned_Full_Data_V6.csv"
df_bridge = pd.read_csv(bridge_data_path, dtype=str,usecols=['STRUCTURE_NUMBER_008','LAT_016','LONG_017'])
```

```python
pred_data_path = "D:\\Bridge_Data\\CSV\\Bridge_Predictions.csv"
df_preds = pd.read_csv(pred_data_path)
```
