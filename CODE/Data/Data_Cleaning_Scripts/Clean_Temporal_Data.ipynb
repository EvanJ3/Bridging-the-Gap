```python
import numpy as np
import csv
import os
import pandas as pd
from filter_scripts import *
from column_conversion_dicts import *
```

```python
#Set TSV file paths and set path to 2020 txt file
tsv_file_path = 'D:\\Bridge_Data\\Highway_Bridges_Data_tsv\\'
indv_path = 'Highway_Bridges_2020.txt'
```

```python
#Load the Uncleaned 2020 dataframe apply cursory filtering and cleaning
df_2020 = pd.read_csv(tsv_file_path+indv_path, sep=',',dtype='str',usecols=use_cols_2020)
df_2020['STRUCTURE_NUMBER_008'] = df_2020.apply(lambda x:  x['STATE_CODE_001']+'_'+ x['STRUCTURE_NUMBER_008'],axis=1)
df_2020 = filter_sci_notation(df_2020,col_name='STRUCTURE_NUMBER_008')
df_2020 = remove_duplicate_structs(df_2020,struct_col_name='STRUCTURE_NUMBER_008',year_col_name='YEAR_BUILT_027')
```

```python
#Load the merged time series dataframe
df_ts = pd.read_csv('D:\\Bridge_Data\\Utilities\\all_merged_time_series.csv', sep=',',dtype='str')
df_ts.drop(['Unnamed: 0'],axis=1,inplace=True)
```

```python
#Merge the two frames
total_merge = pd.merge(df_2020, df_ts, left_on="STRUCTURE_NUMBER_008",right_on="STRUCTURE_NUMBER",how='left')
total_merge = total_merge[total_merge['COUNTY_CODE_003'].notna()]
```
