```python
def long_cleaning_function(x):
    magnitudes = np.array([1,10,100,1000,10000,100000,1000000,10000000,100000000,1000000000])
    if abs(x/1e6)>66.55 and abs(x/1e6)<124.26:
        x_new =  -(int(abs(x)/1e6) + int(abs(x)%1e6/1e4)/60 + abs(x)%1e4/36e4)
    else:
        x_new = x
    
    if x_new > 0.00001:
        x_new = -x_new
    
    
    diff1 = (-125.0 - x_new)
    div_out = diff1/magnitudes
    div_out1 = np.where(div_out<125.0,1,0)
    div_out2 = np.where(div_out>65.0,1,0)
    div_order_idx = np.argmax(div_out1+div_out2)
    mag_i = magnitudes[div_order_idx]
    x_new = x_new/mag_i
    return x_new
```

```python
#def lat_cleaning_function(x):
#    if abs(x/1e6)>24.21 and abs(x/1e6)<49:
#        return int(abs(x)/1e6) + int(abs(x)%1e6/1e4)/60 + abs(x)%1e4/36e4
#    else:
#        return x

def lat_cleaning_function(x):
    
    if abs(x/1e6)>24.21 and abs(x/1e6)<49:
        x_new = int(abs(x)/1e6) + int(abs(x)%1e6/1e4)/60 + abs(x)%1e4/36e4
    else:
        x_new = x
    if x_new < 0.00000:
        x_new = -x_new

    magnitudes = np.array([1,10,100,1000,10000,100000,1000000,10000000,100000000,1000000000])

    diff1 = (x_new-60.0)
    div_out = diff1/magnitudes
    div_out1 = np.where(div_out<60.0,1,0)
    div_out2 = np.where(div_out>20.0,1,0)
    div_order_idx = np.argmax(div_out1+div_out2)
    mag_i = magnitudes[div_order_idx]
    x_new = x_new/mag_i

    return x_new
```

```python
df_merged['LONG_New'] = df_merged.apply(lambda x: long_cleaning_function(float(x['LONG_017'])),axis=1)
df_merged['LAT_New'] = df_merged.apply(lambda x: lat_cleaning_function(float(x['LAT_016'])),axis=1)

df_merged = df_merged[df_merged['LAT_New'] != 0.0]
df_merged = df_merged[df_merged['LAT_New'] >= 20.0]
df_merged = df_merged[df_merged['LAT_New'] <= 60.0]
df_merged = df_merged[df_merged['LONG_New'] <= 0.0]
df_merged = df_merged[df_merged['LONG_New'] <= -60.0]
df_merged = df_merged[df_merged['LONG_New'] >= -130.0]
```

```python
df_merged.drop(columns=['LAT_016','LONG_017'],inplace=True)
df_merged = df_merged.rename(columns={'LONG_New': 'LONG_017', 'LAT_New': 'LAT_016'})
```
