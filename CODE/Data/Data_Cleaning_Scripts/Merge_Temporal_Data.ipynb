```python
tsv_file_path = 'D:\\Bridge_Data\\Highway_Bridges_Data_tsv\\'
initial_dir = os.getcwd()
os.chdir(tsv_file_path)
tsv_file_list = os.listdir()
os.chdir(initial_dir)
```

```python
year_string = tsv_file_list[-1][-8:][:-4]
df_2020 = pd.read_csv(tsv_file_path+tsv_file_list[-1], sep=',',dtype='str')
df_2020 = filter_ts_cols(df_2020)
df_2020['STRUCTURE_NUMBER_008'] = df_2020.apply(lambda x:  x['STATE_CODE_001']+'_'+ x['STRUCTURE_NUMBER_008'],axis=1)
df_2020 = filter_sci_notation(df_2020,col_name='STRUCTURE_NUMBER_008')
df_2020 = remove_duplicate_structs(df_2020,struct_col_name='STRUCTURE_NUMBER_008',year_col_name='YEAR_BUILT_027')
integer_col_names = get_integer_cols(df_2020)
eval_col_names = get_eval_cols(df_2020)
df_2020 = clean_eval_decimal_cols(df_2020,integer_col_names)
df_2020 = clean_eval_codes(df_2020,eval_col_names)
df_2020 = clean_scour_critical_codes(df_2020,col_name='SCOUR_CRITICAL_113')
df_2020 = format_column_names(df_2020,year_string)
df_merged = df_2020
```

```python
df_2020
```

```python
for i in range(0,len(tsv_file_list)-1):
    year_string = tsv_file_list[i][-8:][:-4]
    df = pd.read_csv(tsv_file_path+tsv_file_list[i], sep=',',dtype='str')
    df = filter_ts_cols(df)
    df['STRUCTURE_NUMBER_008'] = df.apply(lambda x:  x['STATE_CODE_001']+'_'+ x['STRUCTURE_NUMBER_008'],axis=1)
    df = filter_sci_notation(df,col_name='STRUCTURE_NUMBER_008')
    df = remove_duplicate_structs(df,struct_col_name='STRUCTURE_NUMBER_008',year_col_name='YEAR_BUILT_027')
    integer_col_names = get_integer_cols(df)
    eval_col_names = get_eval_cols(df)
    df = clean_eval_decimal_cols(df,integer_col_names)
    df = clean_eval_codes(df,eval_col_names)
    df = clean_scour_critical_codes(df,col_name='SCOUR_CRITICAL_113')
    df = format_column_names(df,year_string)
    year_col_string = 'YEAR_BUILT_'+year_string
    df = df.drop([year_col_string], axis=1)
    state_col_string = 'STATE_CODE_'+year_string
    df = df.drop([state_col_string], axis=1)
    right_join_string = 'STRUCTURE_NUMBER_'+year_string
    df_merged = pd.merge(df_merged, df, left_on="STRUCTURE_NUMBER_2020",right_on=right_join_string,how='left')
    structure_col_string = 'STRUCTURE_NUMBER_'+year_string
    df_merged = df_merged.drop([structure_col_string], axis=1)
    print(f'Merge Number {i}  Year {year_string} Completed')
    print(f'Current Frame Size = {df_merged.shape}')
    print('----------------------------------------')
```

```python
df_merged.to_csv('time_series_merged_data.csv')
```
